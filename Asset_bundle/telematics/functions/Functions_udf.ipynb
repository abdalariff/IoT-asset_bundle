```python
%sql
CREATE OR REPLACE FUNCTION IsOverSpeeding(
    Speed FLOAT,
    SpeedLimit FLOAT
)
RETURNS STRING
RETURN CASE WHEN Speed > SpeedLimit THEN 'Yes' ELSE 'No' END;
```

```python
%sql
CREATE OR REPLACE FUNCTION IsLowFuel(
    FuelLevel FLOAT
)
RETURNS STRING
RETURN CASE
    WHEN FuelLevel < 10 THEN 'Critical'
    WHEN FuelLevel < 25 THEN 'Low'
    ELSE 'Normal'
END;
```

```python
%sql
CREATE OR REPLACE FUNCTION TirePressureStatus(
    FL FLOAT,
    FR FLOAT,
    RL FLOAT,
    RR FLOAT
)
RETURNS STRING
RETURN CASE
    WHEN FL < 28 OR FL > 36 OR FR < 28 OR FR > 36
      OR RL < 28 OR RL > 36 OR RR < 28 OR RR > 36
    THEN 'Alert'
    ELSE 'Normal'
END;
```

```python
%sql
CREATE OR REPLACE FUNCTION EngineTempStatus(
    EngineTemp FLOAT
)
RETURNS STRING
RETURN CASE
    WHEN EngineTemp > 110 THEN 'High'
    WHEN EngineTemp < 70 THEN 'Low'
    ELSE 'Normal'
END;
```

```python
%sql
CREATE OR REPLACE FUNCTION VehicleRiskScore(
    Speed FLOAT,
    FuelLevel FLOAT,
    TirePressureAlert BOOLEAN,
    EngineTemp FLOAT,
    GeoFenceStatus BOOLEAN
)
RETURNS INT
RETURN LEAST(
    CASE WHEN Speed > 100 THEN 50 ELSE 0 END
  + CASE WHEN FuelLevel < 15 THEN 20 ELSE 0 END
  + CASE WHEN TirePressureAlert THEN 30 ELSE 0 END
  + CASE WHEN EngineTemp > 110 THEN 20 ELSE 0 END
  + CASE WHEN GeoFenceStatus = FALSE THEN 25 ELSE 0 END
  , 100
);
```
