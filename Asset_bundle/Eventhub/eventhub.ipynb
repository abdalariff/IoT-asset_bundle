```python
from pyspark.sql import SparkSession
from pyspark.sql.types import *
from datetime import datetime
import random
import time

spark = SparkSession.builder \
    .appName("FleetSensorSimulation") \
    .getOrCreate()

schema = StructType([
    StructField("EventID", IntegerType(), True),
    StructField("VehicleID", IntegerType(), True),
    StructField("Timestamp", TimestampType(), True),
    StructField("Latitude", FloatType(), True),
    StructField("Longitude", FloatType(), True),
    StructField("Speed", FloatType(), True),
    StructField("FuelLevel", FloatType(), True),
    StructField("EngineTemp", FloatType(), True),
    StructField("Odometer", FloatType(), True),
    StructField("GeoFenceID", IntegerType(), True),
    StructField("GeoFenceStatus", BooleanType(), True),
    StructField("DoorStatus", BooleanType(), True),
    StructField("AuxEquipment1Status", BooleanType(), True),
    StructField("AuxEquipment2Status", BooleanType(), True),
    StructField("AuxEquipment3Status", BooleanType(), True),
    StructField("AuxEquipment4Status", BooleanType(), True),
    StructField("BatteryVoltage", FloatType(), True),
    StructField("TirePressureFrontLeft", FloatType(), True),
    StructField("TirePressureFrontRight", FloatType(), True),
    StructField("TirePressureRearLeft", FloatType(), True),
    StructField("TirePressureRearRight", FloatType(), True),
    StructField("Temperature", FloatType(), True),
    StructField("Humidity", FloatType(), True),
])

number = 1
while True:
    data = [(number,
             random.randint(1, 50),
             datetime.now(),
             round(random.uniform(24, 37), 6),
             round(random.uniform(66, 77), 6),
             round(random.uniform(0, 120), 2),
             round(random.uniform(0, 100), 2),
             round(random.uniform(70, 120), 2),
             round(random.uniform(0, 500000), 2),
             random.randint(1, 10),
             bool(random.randint(0, 1)),
             bool(random.randint(0, 1)),
             bool(random.randint(0, 1)),
             bool(random.randint(0, 1)),
             bool(random.randint(0, 1)),
             bool(random.randint(0, 1)),
             round(random.uniform(11, 14), 2),
             round(random.uniform(28, 36), 2),
             round(random.uniform(28, 36), 2),
             round(random.uniform(28, 36), 2),
             round(random.uniform(28, 36), 2),
             round(random.uniform(-10, 50), 2),
             round(random.uniform(10, 90), 2))]

    df = spark.createDataFrame(data, schema=schema)

    df.write.format("delta").mode("append").saveAsTable("FleetSensorEvents")

    number += 1
    time.sleep(1)
```
